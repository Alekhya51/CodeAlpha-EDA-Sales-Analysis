# TREND ANALYSIS

#### 11.Which month generated the highest sales?

```python
df["Sales_Date"] = pd.to_datetime(df["Sales_Date"])

df["Month"] = df["Sales_Date"].dt.month_name()

monthly_sales = df.groupby("Month")["Sales_Amount"].sum().sort_values(ascending=False)

monthly_sales.head(1)
```

```python
import matplotlib.pyplot as plt
monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(8,5)
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.show()
```

#### 12.Which quarter generated the highest sales?

```python
df["Quarter"] = df["Sales_Date"].dt.quarter

quarter_sales = df.groupby("Quarter")["Sales_Amount"].sum().sort_values(ascending=False)

quarter_sales.head(1)
```

```python
quarter_sales.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Quarter-wise Sales")
plt.xlabel("Quarter")
plt.ylabel("Sales Amount")
plt.show()
```

# Hypothesis Testing and Validation Using Statistics and Visualization

### Hypothesis 1: 
Employees with more experience generate higher sales.

#### * Statistical Analysis

```python
corr = df["Experience_Years"].corr(df["Sales_Amount"])

print("Correlation:", corr)
```

#### * Visualization

```python


plt.figure(figsize=(8,5))
plt.scatter(df["Experience_Years"], df["Sales_Amount"])

plt.title("Experience vs Sales")
plt.xlabel("Experience Years")
plt.ylabel("Sales Amount")

plt.show()
```

Correlation = 0.03
Employee experience has very little impact on sales performance.

### Hypothesis 2:
Some departments perform better than others.

#### * Statistical Analysis

```python
dept_avg = df.groupby("Department")["Sales_Amount"].mean()

print(dept_avg.sort_values(ascending=False))
```

#### * Visualization 

```python
dept_avg.sort_values(ascending=False).plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Sales by Department")
plt.xlabel("Department")
plt.ylabel("Average Sales")

plt.show()
```

Departments do not contribute equally. The department with the highest average sales performs best.

## Hypothesis 3:
Sales performance varies across regions.

#### * Statistical Analysis

```python
region_sales = df.groupby("Region")[
    "Sales_Amount"
].sum()

print(region_sales.sort_values(ascending=False))
```

#### * Visualization

```python
region_sales.sort_values(
    ascending=False
).plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales Amount")

plt.show()
```

Certain regions contribute more revenue than others.

## Hypothesis 4:
Most employees achieve their sales targets.

#### * Statistical 

```python
target_achieved = (
    df["Sales_Amount"] >= df["Target_Sales"]
)

print(target_achieved.value_counts())
```

#### * Visualization

```python
target_achieved.value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Target Achievement Status")

plt.ylabel("")

plt.show()
```

A majority of employees achieved their sales targets.
